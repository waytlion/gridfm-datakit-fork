```python
# Calculate mean load per bus
mean_load_per_bus = df.groupby("bus_id")["load"].mean()

# Count buses with positive and negative mean load
positive_load_buses = (mean_load_per_bus > 0).sum()
negative_load_buses = (mean_load_per_bus < 0).sum()
zero_load_buses = (mean_load_per_bus == 0).sum()

print(f"Buses with positive mean load: {positive_load_buses}")
print(f"Buses with negative mean load: {negative_load_buses}")
print(f"Buses with zero mean load: {zero_load_buses}")
print(f"Total buses: {len(mean_load_per_bus)}")

# Show the distribution 
print("\nMean load per bus:")
print(mean_load_per_bus.sort_values(ascending=False))
```
